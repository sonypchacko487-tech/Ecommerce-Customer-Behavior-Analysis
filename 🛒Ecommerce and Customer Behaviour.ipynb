 🧹 Data Cleaning & Pre-processing of E-Commerce Customer Dataset

### Data Cleaning

- Missing values were handled using appropriate methods
- Data types were verified
- Dataset prepared for analysis

This ensures accurate and reliable insights.

```python
# Create a copy for cleaning
df_clean = df.copy()
```

```python
# Standardize column names (remove spaces and special characters)
df_clean.columns = df_clean.columns.str.strip().str.replace(' ', '_').str.replace('-', '_')
print("Column names standardized:", list(df_clean.columns))
```

## 🕳️ Handle Missing Values

```python
# Numerical columns - fill with median 
numerical_cols = df_clean.select_dtypes(include=np.number).columns
for col in numerical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Categorical columns - fill with mode
categorical_cols = df_clean.select_dtypes(include='object').columns
for col in categorical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Missing values after handling:")
df_clean.isnull().sum()
```

## Final Data Quality Check

```python
print("Final dataset info:")
df_clean.info()
print(f"Remaining missing values: {df_clean.isnull().sum().sum()}")
```

##  💾 Save the Cleaned Dataset

```python
df_clean.to_csv("cleaned_ecommerce_customer_behaviour_dataset.csv", index=False)
print("Cleaned dataset saved successfully!")
```

### 🔹 Pivot Table: 1. Purchases by Gender and Churn

```python
pivot1 = pd.pivot_table(
    df_clean,
    values='Total_Purchases',
    index='Gender',
    columns='Churned',
    aggfunc='mean'
)

pivot1
```

What this shows:
This pivot table summarizes the average number of purchases based on:
- Gender (rows)
- Churn status (0 = Active, 1 = Churned) (columns)
- Across rows → Male vs Female
- Across columns → Churned vs Active
Example:
- If value under 0 (Active) is higher → active users purchase more
- If value under 1 (Churned) is lower → churned users had low engagement

Insight:
- Average purchases are similar across genders, suggesting purchasing behavior is not gender-dependent.

### Purchases by Country

```python
pivot2 = pd.pivot_table(
    df_clean,
    values='Total_Purchases',
    index='Country',
    aggfunc='mean'
).sort_values(by='Total_Purchases', ascending=False)

pivot2
```

Insight:
- Customers from certain countries show higher purchasing activity, indicating potential target markets.
